```python
from astropy.io import fits
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from GalVel_Pop import gal_population, UVW
```

Reading the APOGEE DR17 data 

```python
from astropy.table import Table
dat = Table.read('APOKASC_cat_v7.4.0.fits', format='fits')
names = [name for name in dat.colnames if len(dat[name].shape) <= 1]
allstar = dat[names].to_pandas()
```

```python
allstar.columns.tolist()
```

```python
apocasc3_ages = allstar[allstar['APOKASC3_AGE_RCAGB'] > 0]
apocasc3_ages['GAIAEDR3_SOURCE_ID'] = apocasc3_ages['GAIAEDR3_SOURCE_ID'].astype(str)
apocasc3_ages['APOKASC3_CAT_AGE'] = apocasc3_ages['APOKASC3_CAT_AGE'].astype(str)
```

```python
columns = ['DR17_FE_H', 'DR17_VHELIO_AVG', 'APOKASC3_CAT_AGE', 'GAIAEDR3_PARALLAX',
 'GAIAEDR3_PARALLAX_ERROR',
 'GAIAEDR3_PMRA',
 'GAIAEDR3_PMRA_ERROR',
 'GAIAEDR3_PMDEC',
'GAIAEDR3_PMDEC_ERROR',
'GAIAEDR3_PHOT_G_MEAN_MAG',
 'GAIAEDR3_PHOT_BP_MEAN_MAG',
 'GAIAEDR3_PHOT_RP_MEAN_MAG',
 'RA',
 'DEC', 'GAIAEDR3_SOURCE_ID', 'APOKASC3_AGE_RGB', 'APOKASC3_AGE_RCAGB']
```

```python
apocasc3_ages_restricted = apocasc3_ages[columns]
```

```python
column_rename_map = {
    'DR17_FE_H': 'fe_h',
    'DR17_VHELIO_AVG': 'radial_velocity',
    'GAIAEDR3_PARALLAX': 'parallax',
    'GAIAEDR3_PARALLAX_ERROR': 'parallax_error',
    'GAIAEDR3_PMRA': 'pmra',
    'GAIAEDR3_PMRA_ERROR': 'pmra_error',
    'GAIAEDR3_PMDEC': 'pmdec',
    'GAIAEDR3_PMDEC_ERROR': 'pmdec_error',
    'GAIAEDR3_PHOT_G_MEAN_MAG': 'phot_g_mean_mag',
    'GAIAEDR3_PHOT_BP_MEAN_MAG': 'phot_bp_mean_mag',
    'GAIAEDR3_PHOT_RP_MEAN_MAG': 'phot_rp_mean_mag',
    'RA': 'ra',
    'DEC': 'dec',
    'GAIAEDR3_SOURCE_ID': 'source_id',
    'APOKASC3_AGE_RGB': 'age_rgb',
    'APOKASC3_AGE_RCAGB': 'age_rcagb'
}

apocasc3_ages_restricted.rename(columns=column_rename_map, inplace=True)
```

```python
apocasc3_ages_restricted
```

```python
# Define a function to apply gal_population to each row
def calculate_UVW(row):
    UVW_values =  UVW(row['ra'], row['dec'], row['pmra'], row['pmdec'], row['radial_velocity'], row['parallax'])
    return pd.Series(UVW_values, index=['U_lsr', 'V_lsr', 'W_lsr'])

apocasc3_ages_restricted[['U_lsr', 'V_lsr', 'W_lsr']] = apocasc3_ages_restricted.apply(lambda row: calculate_UVW(row), axis=1)
```

```python
# Define a function to apply gal_population to each row
def calculate_probabilities(row):
    probs =  gal_population(row['U_lsr'], row['V_lsr'], row['W_lsr'], char_vel='Mean')
    return probs[0]

# Apply the function to each row using apply with axis=1
apocasc3_ages_restricted['P_gal'] = apocasc3_ages_restricted.apply(calculate_probabilities, axis=1)
```

```python
apocasc3_ages_restricted.dropna(inplace=True)
apocasc3_ages_restricted.shape
```
