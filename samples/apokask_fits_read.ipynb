Reading the APOGEE DR17 data 

```python
from astropy.table import Table
dat = Table.read('APOKASC_cat_v7.4.0.fits', format='fits')
names = [name for name in dat.colnames if len(dat[name].shape) <= 1]
allstar = dat[names].to_pandas()
```

```python
apocasc3_ages = allstar[allstar['APOKASC3_AGE_RCAGB'] > 0]
apocasc3_ages['GAIAEDR3_SOURCE_ID'] = apocasc3_ages['GAIAEDR3_SOURCE_ID'].astype(int)
apocasc3_ages['APOKASC3_CAT_AGE'] = apocasc3_ages['APOKASC3_CAT_AGE'].astype(str)
```

```python
columns = ['DR17_FE_H', 'APOKASC3_CAT_AGE', 'GAIAEDR3_PARALLAX',
 'GAIAEDR3_PARALLAX_ERROR',
 'GAIAEDR3_PMRA',
 'GAIAEDR3_PMRA_ERROR',
 'GAIAEDR3_PMDEC',
'GAIAEDR3_PMDEC_ERROR',
'GAIAEDR3_PHOT_G_MEAN_MAG',
 'GAIAEDR3_PHOT_BP_MEAN_MAG',
 'GAIAEDR3_PHOT_RP_MEAN_MAG',
 'RA',
 'DEC', 'GAIAEDR3_SOURCE_ID', 'APOKASC3_AGE_RGB', 'APOKASC3_AGE_RCAGB']
```

```python
apocasc3_ages_final = apocasc3_ages[columns]
apocasc3_ages_final.to_csv('apocasc3_ages_final.csv', index=False)
```
